```python
from nltk.util import trigrams, ngrams
```

```python
str = " generative ai and agentic ai is automate the industry and more focus on every domain"
quotes_token = nltk.word_tokenize(str)
quotes_token
```

```python
len(quotes_token)
```

```python
quotes_biigrams = list(nltk.bigrams(quotes_token))
quotes_biigrams 
```

```python
quotes_trigrams = list(nltk.trigrams(quotes_token))
quotes_trigrams             
```

```python
quotes_ngrams = list(nltk.ngrams(quotes_token))
quotes_ngrams 
```

```python
quotes_ngrams = list(nltk.ngrams(quotes_token , 5))
quotes_ngrams 
```

## Stemming --> it do suffix stripping 


```python
from nltk.stem import PorterStemmer # doesn't work for all 
pst = PorterStemmer()
```

```python
pst.stem('affection')
```

```python
pst.stem('playing')
```

```python
pst.stem('maximum')
```

```python
words_to_stem =['give','giving','given','gaved','thinking','loving','maximum','Hello']

for words in words_to_stem:
    print(words+ ':' + pst.stem(words))
```

```python
from nltk.stem import LancasterStemmer  # sufffix stripping for all words
lst = LancasterStemmer()

for word in words_to_stem:
    print(word+ ':' + lst.stem(word))
    
```

```python
from nltk.stem import SnowballStemmer # mainly works as porter stemmer
sbst = SnowballStemmer('english')

for word in words_to_stem:
    print(word+ ':' + sbst.stem(word))
```
